# BUSINESS OBJECTIVE

This project aims to move beyond traditional churn prediction by identifying which customers are most valuable to retain. The objective is to analyze customer behavior, predict churn risk, and quantify the financial impact of churn using a value-based approach.

Specifically, this project seeks to:

*   Analyze customer data to uncover key drivers of churn through exploratory data analysis (EDA)
*   Develop and evaluate predictive models (Logistic Regression, Random Forest, XGBoost, CatBoost) to estimate churn probability
*   Interpret model outputs using Feature Importance and SHAP to understand key influencing factors
*   Introduce a value-based prioritization metric (Churn Probability × Customer Value) to estimate revenue at risk
*   Design and simulate targeted retention strategies, comparing mass campaigns vs precision targeting to maximize ROI

# MODEL CREATION AND EVALUATION


Now, let's move on to the predictive models. In this notebook, we will use the Area Under the Curve of Receiver Characteristic Operator (AUC-ROC or ROC-AUC) as the main metric to assess the performance of our models. The ROC-AUC measures the ability of a model is to distinguish between classes. [(Link for more information about ROC-AUC)](https://www.analyticsvidhya.com/blog/2020/06/auc-roc-curve-machine-learning/). Nevertheless, we will also check the accuracy and the F1-Score, plus the classification report and the confusion matrix for each model.

First, we will make a copy of the dataset and separate the features from the target.

```python
X = df.copy().drop('Churn', axis = 1)
Y = df['Churn'].copy()
```

We can use the Column Transformer to assign each transformation to its correct features and fit it in a pipeline as a preprocessing step.

```python
num_features = num_accinfo_features

cat_3p_features = []
for col in X.columns:
    if (X[col].nunique() > 2) & (X[col].nunique() < 5):  #less than 5 to exclude the numerical features
        cat_3p_features.append(col)
print('Numerical features: ', num_features, '\n')
print('Nominal with 3 or more categories: ', cat_3p_features)
```

Even without the intent of doing an extensive hyperparameter tuning, we can give each model a better chance of good performance by testing some values for a key parameter and choosing one of them based on cross-validation score.

```python
def cv_function (model, param, list):

    rp_st_kfold = RepeatedStratifiedKFold(n_splits=10, n_repeats = 3, random_state = 42)
    search_model = model
    print ('Hyperparameter: ', param)

    for i in list:
        param_dict = {param : i}
        search_model.set_params(**param_dict)
        cv_score = cross_val_score(search_model, X_train, Y_train, cv=rp_st_kfold, scoring='roc_auc')
        print("Parameter: {0:0.2f} - AUC(SD): {1:0.4f} ({2:0.4f})". format(i, cv_score.mean(), cv_score.std()))

params_lr_list = [0.01,0.1,0.2,0.3,0.5,0.7,1,2,3,5]
param_lr = 'LR__C'
cv_function(lr_pipe, param_lr, params_lr_list)
```

## <a id="47">Class Imbalance</a>

As shown in the beginning of the EDA stage, the dataset is imbalanced. 26,5% of the samples represent customers who left in the last month. Such imbalance is somewhat expected, given the nature of this problem, but it comes with issue that the predictions will be biased towards the majority class.
Neither one of the models were able to reach a recall of 56%+, which means that if we were to use the ‘hard’ (0 or 1) predictions to target customers that were close to churn, we would miss almost half of them. There are a few options to deal with class imbalance, such as:

-	Apply Oversampling/Undersampling: Generate new samples for the minority class or remove some samples from the majority class in order to have a balanced dataset;
-	Tune the ‘class weighting’ hyperparameter: Some models have a specific hyperparameter (e.g.: XGBoost’s ‘scale_pos_weight’) which can be used to assign a higher weight to the misclassification of a minority class sample;
-	Adjust the classification threshold: The default threshold for binary classification models is 50%. Instead of using the hard predictions, we can get the probabilities of churn for each sample and classify them based on a custom threshold.

The last technique, although simple, can be quite effective depending on the goal. We can tune the threshold, for instance, to optimize the F1 score or to find the best recall at a minimum expected value for precision.


This notebook will be updated on a near feature to present how we can optimize the threshold. For now, I’ll be showing a quick example of how using a custom threshold can affect the classification metrics.

Let’s get the probabilities found by the CatBoost and classify the test set using a threshold of 25%.

```python
custom_th_pred= np.where(cat_pred_proba>0.250, 1, 0)
print(metrics.classification_report(Y_test, custom_th_pred, labels = [0, 1], digits = 3))
```

We were able to find a recall of 82.4%, against less than 56% from the original predictions. Of course, there is a tradeoff involved. Reducing the threshold comes with the downside of losing precision. Yet, the F1-score was also better with the custom threshold (0.634 vs 0.611).

```python

# Define thresholds
value_threshold = impact_df['MonthlyCharges'].median()
risk_threshold = 0.6

# Create segments
impact_df['Segment'] = 'Other'

impact_df.loc[
    (impact_df['Churn_Probability'] > risk_threshold) &
    (impact_df['MonthlyCharges'] >= value_threshold),
    'Segment'
] = 'High-Value High-Risk'

impact_df.loc[
    (impact_df['Churn_Probability'] > risk_threshold) &
    (impact_df['MonthlyCharges'] < value_threshold),
    'Segment'
] = 'Low-Value High-Risk'

impact_df.loc[
    (impact_df['Churn_Probability'] <= risk_threshold) &
    (impact_df['MonthlyCharges'] >= value_threshold),
    'Segment'
] = 'High-Value Low-Risk'


```

# MASS CAMPAIGN

```python

def simulate_mass_campaign(df, prob_threshold=0.6, discount_rate=0.20, success_rate=0.15):
    mass_targets = df[df['Churn_Probability'] > prob_threshold].copy()

    total_cost = (mass_targets['MonthlyCharges'] * discount_rate).sum()

    revenue_saved = (
        mass_targets[mass_targets['Actual_Churn'] == 1]['MonthlyCharges'] * success_rate
    ).sum()

    net_profit = revenue_saved - total_cost
    roi = (net_profit / total_cost * 100) if total_cost > 0 else 0

    return net_profit, roi, len(mass_targets)
```

#PRECISION CAMPAIGN (Optimized strategy)

```python

def simulate_precision_campaign(df, discount_rate=0.10, success_rate=0.25):
    targeted = df[df['Target_for_Campaign'] == 1].copy()

    total_cost = (targeted['MonthlyCharges'] * discount_rate).sum()

    revenue_saved = (
        targeted[targeted['Actual_Churn'] == 1]['MonthlyCharges'] * success_rate
    ).sum()

    net_profit = revenue_saved - total_cost
    roi = (net_profit / total_cost * 100) if total_cost > 0 else 0

    return net_profit, roi, len(targeted)
```

```python
# --- 3. RUN BOTH STRATEGIES ---
mass_profit, mass_roi, mass_count = simulate_mass_campaign(impact_df)
prec_profit, prec_roi, prec_count = simulate_precision_campaign(impact_df)
```

```python
# --- 4. COMPARISON TABLE (FOR POWER BI / VISUALS) ---
comparison_df = pd.DataFrame({
    'Campaign': ['Mass Campaign', 'Precision Strategy'],
    'Target_Customers': [mass_count, prec_count],
    'Net_Profit': [mass_profit, prec_profit],
    'ROI (%)': [mass_roi, prec_roi]
})

print("\n===== CAMPAIGN COMPARISON =====")
print(comparison_df)
```

The initial simulation shows a negative ROI, suggesting that the cost of offering discounts to all predicted churners exceeds the recovered revenue. This occurs because the model, while high in recall, includes many "False Positives."

To fix this, we will shift from a "Broad Discount" strategy to a "Precision Targeting" strategy. We will only target the High-Value at Risk segment (high probability + high monthly charges) and adjust the threshold to prioritize Precision over pure Recall.

We reduced the retention offer from a standard 20% to a targeted 10%. Because we are focusing only on our highest spenders, a 10% loyalty discount is enough to trigger a 'stay' decision without unnecessarily eroding our profit margins
